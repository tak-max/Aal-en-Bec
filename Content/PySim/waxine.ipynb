# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

Voor optie 1 zouden we verwachten dat we de volgende waarnemingen zullen doen. Op het moment dat de cilinder over het kaarsje wordt gezet, zal het kaarsje langzaam gaan stijgen door het stijgende water. De snelheid van het stijgen zal evenredig zijn met de grootte van de vlam, en het stijgen zal direct stoppen als de kaars uit is. Reden hiervoor is dat het zuurstof geleidleijk wordt verbrand. 

Voor optie 2 verwachten we waar te nemen dat doordat het gas uitzet door de verwarming het volume toeneemt en hierdoor een deel van het gas ontsnapt in de vorm van bubbels, vervolgens zal pas als de kaars uit is het volume afnemen doordat de lucht in de cilinder afkoetl. De onderdruk die hierdoor ontstaat zal het water ophoog trekken en zal het water dus gaan stijgen. 

Voor optie 3 zouden we verwachten dat er condens aan de binnenzijde van de cilinder ontstaat gedurende de verbranding en vlak erna. Stijging van het waterniveau zal plaatsvinden tijdens de verbranding. 


Waarnemingen: vlak na het plaatsen van de cilinder komen er bubbels vanuit de cilinder vandaan, dit gebeurd gedurende het branden van het kaarsje, voornamelijk aan het begin van het branden. Op het moment dat het kaarsje bijna uit is, begint het waterniveau te stijgen. Het waternivevau stijgt sneller naarmate de kaars helemaal uit is gegaan. Ook is er aan de onderkant van de cilinder een kleine hoeveelheid condensatie zichtbaar. 


Conclusie:
